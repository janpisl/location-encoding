```python
driver_classes = np.unique(gdf.driver, return_counts=True)[0]
max_count_per_class = 5000
def flatten(nested_list):
    return [x for xs in nested_list for x in xs]


indices = []
for _class in driver_classes:
    try:
        class_samples = gdf.loc[gdf.driver == _class].sample(max_count_per_class, random_state=1).index
    except ValueError:
        class_samples = gdf.loc[gdf.driver == _class].index
    indices.append(class_samples.tolist())

indices = flatten(indices)

subset = gdf.loc[indices]
```

```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import box

def calculate_cell_density(gdf, width, height, bbox):
    # Load landmass data
    world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))

    # Create the grid of cells
    minx, miny, maxx, maxy = bbox
    rows = int((maxy - miny) / height)
    cols = int((maxx - minx) / width)
    cells = []

    for i in range(cols):
        for j in range(rows):
            lower_left = (minx + i * width, miny + j * height)
            upper_right = (lower_left[0] + width, lower_left[1] + height)
            cells.append(box(lower_left[0], lower_left[1], upper_right[0], upper_right[1]))

    cell_gdf = gpd.GeoDataFrame(geometry=cells)
    
    # Calculate the actual land area for each cell
    cell_land_areas = []
    for cell in cell_gdf.geometry:
        # Intersect cell with landmass polygons and sum their areas
        intersected_areas = world[world.intersects(cell)].geometry.intersection(cell).area
        cell_land_areas.append(intersected_areas.sum())

    cell_gdf['land_mass'] = cell_land_areas

    # Calculate the number of points in each cell
    cell_gdf['point_count'] = cell_gdf.geometry.apply(lambda x: gdf.within(x).sum())

    # Calculate the density of points per unit of land mass
    # Avoid division by zero
    cell_gdf['density'] = cell_gdf.apply(lambda row: row['point_count'] / row['land_mass'] if row['land_mass'] > 0 else 0, axis=1)

    return cell_gdf
```

```python
densities = calculate_cell_density(gdf, 2, 2, (-50, -6, -40, 6))
```

```python
_densities = densities.loc[densities.land_mass > 0]
```

```python
_densities = _densities.loc[densities.point_count > 0]
```
