

### Goal
- find locations of forest loss and their drivers

### inputs
- Vancutsem Tropical Moist Forest (TMF) deforestation year
- MapBiomas
- class indices in MapBiomas which i want to recognize 


### method
- iterate by 1km squares through the Hansen tiles
- if any forest loss (or more than threshold)
    - extract the geometry
    - extract the underlying drivers

           
           
### output 
- gdf
    - columns
        - sampleid
        - geometry (1km2) 
        - amount of forest loss in geometry (in pixels or area or %) 
        - for each class, the amount of forest loss attributed to that class (incl. "Other")
        - for each year in target period, the amount of forest loss in that year
        - main forest loss year (in the final version, this should also be per-pixel) 
        - main driver class
            - in the final version, let's keep the per-pixel labels 
                - that way, the main driver class is just the idxmax
                - i can also know 'how main' the main driver is (if its 100% or e.g. 30%)
                - and its would be a segmentation-level dataset

```python
import os
from functools import partial
from timeit import default_timer as timer
from datetime import timedelta
import ast

import rasterio
from rasterio.mask import mask
from rasterio.features import shapes
import geopandas as gpd
from shapely.geometry import shape, Polygon
import pyproj
from shapely.ops import transform
import numpy as np
import pandas as pd
import geopandas as gpd
from rasterio.plot import show
import tqdm
from shapely.geometry import box
from rasterio.errors import RasterioIOError
from rasterio.enums import Resampling
```

```python
def parse_lat(latitude: str) -> tuple:

    assert ("S" in latitude or "N" in latitude), 'Unexpected name format'
    if "S" in latitude:
        max_lat = int(latitude.split("S")[1]) * -1
    else:
        max_lat = int(latitude.split("N")[1])
    min_lat = max_lat - 10

    return min_lat, max_lat

def parse_lon(longitude: str) -> tuple:
    assert ("W" in longitude or "E" in longitude), 'Unexpected name format'
    if "W" in longitude:
        max_lon = int(longitude.split("W")[1]) * -1       
    else:
        max_lon = int(longitude.split("E")[1])    
    min_lon = max_lon + 10

    return min_lon, max_lon

def get_extent(filename):
    _, lat, lon = filename.split('_')
    min_lat, max_lat = parse_lat(lat)
    min_lon, max_lon = parse_lon(lon)

    return min_lat, max_lat, min_lon, max_lon


```

```python
#target_years = [2017,2018,2019,2020,2021]
target_years = list(range(2000, 2020))
```

```python
mapbiomas_classes = {
    #1 : 'forest', # superclass
    #3 : 'forest',  # "Forest Formation"
    #10 : 'non-forest natural formation', #incl. wetland, grassland etc. - I think this is the 'natural regrowth'
    
    # Commodities & drivers
    15 : 'pasture', 
    9 : 'forest plantation',
    39 : 'soy',
    46 : 'coffee',
    35 : 'palm oil',
    30 : 'mining',
    24 : 'urban',
}

```

```python
forestloss_dir = '../../../Vancutsem_yearloss/'



shapes = []
filenames = []
for filename in os.listdir(forestloss_dir):
    if not filename.endswith('.tif'):
        continue

    min_lat, max_lat, min_lon, max_lon = get_extent(filename.split('.')[0])
    polygon = Polygon([(min_lon, min_lat), (min_lon, max_lat), (max_lon, max_lat, ), (max_lon, min_lat )])
    shapes.append(polygon)
    filenames.append(filename)
```

```python
forestloss_tiles = gpd.GeoDataFrame(geometry=shapes)
forestloss_tiles['filename'] = filenames
```

```python
#TODO: currently i am just choosing an arbitrary year
#i should look at the year AFTER the forest loss
mapbiomas_path = '/home/jan/Documents/EPFL/drivers/data/mapbiomas/brasil_coverage_2022.tif'

with rasterio.open(mapbiomas_path) as src:
    mapb_src_profile = src.profile
    mapb_bounds = src.bounds
    
mapbiomas_extent = box(*mapb_bounds)
```

```python
forestloss_brazil_tiles = forestloss_tiles.loc[forestloss_tiles.intersects(mapbiomas_extent)]
```

```python
geoms = []
forest_loss_years = []
drivers = []


for idx, row in forestloss_brazil_tiles.iterrows():
    print(idx)

    path = os.path.join(forestloss_dir, row.filename)

    with rasterio.open(path) as src:
        forestloss_profile = src.profile
        
        
    #Define the window of MapBiomas extent (to use for reading the Hansen tile)
    mapb_window = rasterio.windows.from_bounds(left=mapb_bounds.left, 
                                          bottom=mapb_bounds.bottom, 
                                          right=mapb_bounds.right, 
                                          top=mapb_bounds.top,
                                          transform=forestloss_profile['transform'], 
                                         )
    
    
    #Read the GFC raster using the MapBiomas window
    with rasterio.open(path) as src:
        forestloss_data = src.read(1, window=mapb_window)
        forestloss_profile = src.profile
        forestloss_bounds_orig = src.bounds

    

    #TODO: co to je za nesmysl, tohle je uplne zbytecny
    #e.g., max(forestloss_bounds_orig.left, mapb_bounds.left) ma by primo argument do toho dalsiho window
    #Define a window as the intersection of given Hansen tile and MapBiomas
    forestloss_bounds = rasterio.coords.BoundingBox(
        left=max(forestloss_bounds_orig.left, mapb_bounds.left), 
        bottom=max(forestloss_bounds_orig.bottom, mapb_bounds.bottom), 
        right=min(forestloss_bounds_orig.right, mapb_bounds.right), 
        top=min(forestloss_bounds_orig.top, mapb_bounds.top)
    )

        
    with rasterio.open(mapbiomas_path) as src:
        mapb_profile = src.profile #This is needed to compute the window

    #Read a part of the MapBiomas raster corresponding to the intersection of it with the GFC tile
    window = rasterio.windows.from_bounds(left=forestloss_bounds.left, 
                                          bottom=forestloss_bounds.bottom, 
                                          right=forestloss_bounds.right, 
                                          top=forestloss_bounds.top,
                                          transform=mapb_profile['transform'], 
                                          #height=None, width=None, precision=None
                                         )

        
    with rasterio.open(mapbiomas_path) as src:
        mapb_data = src.read( 
            out_shape=(
                src.count,
                forestloss_data.shape[0],
                forestloss_data.shape[1]
                ),
            resampling=Resampling.bilinear,
            window=window)
        mapb_data = mapb_data[0]
    

    
    if mapb_data.max() == 0: #Means there are no MapBiomas data for the area
        print(f"No MapBiomas data for idx {idx}")
        continue

    rows, cols = forestloss_data.shape
    patch_rows, patch_cols = 35, 35
    

    for i in range(0, rows, patch_rows):
        for j in range(0, cols, patch_cols):
            forestloss_patch = forestloss_data[i:i+patch_rows, j:j+patch_cols]

            mapb_patch = mapb_data[i:i+patch_rows, j:j+patch_cols]
            
            if not np.isin(forestloss_patch, target_years).any(): #No forest loss in target period within the window
                continue
            if mapb_patch.max() == 0: #Means there are no MapBiomas data for the window
                continue
            
            forest_loss_mask = (np.isin(forestloss_patch, target_years))
            
            #lower_left_coords = rasterio.transform.xy(forestloss_profile['transform'], i,j,center='ll')
            #upper_right_coords = rasterio.transform.xy(forestloss_profile['transform'], i+patch_rows,j+patch_cols,center='ur')

            lower_left_coords = rasterio.transform.xy(
                forestloss_profile['transform'], 
                i + round(max(0, mapb_window.row_off)), 
                j + round(max(0, mapb_window.col_off)),  #TODO: test this works as intended
                center='ll')
            upper_right_coords = rasterio.transform.xy(
                forestloss_profile['transform'], 
                i + round(max(0, mapb_window.row_off)) + patch_rows,
                j + round(max(0, mapb_window.col_off)) + patch_cols,  #TODO: test this works as intended
                center='ur')
            
            
            
            _geom = box(lower_left_coords[0], lower_left_coords[1], upper_right_coords[0], upper_right_coords[1])

            _forest_loss_years = np.unique(forestloss_patch[forest_loss_mask], return_counts=True)
            _drivers = np.unique(mapb_patch[forest_loss_mask], return_counts=True)
            
            geoms.append(_geom)
            forest_loss_years.append(_forest_loss_years)
            drivers.append(_drivers)

            
    print(f"Items: {len(geoms)}")
```

```python
#classes, counts = np.unique(drivers, return_counts=True)
#main_driver = classes[counts.argmax()]
#main_driver_ratio = counts[counts.argmax()]/counts.sum() #1 means that it's the only driver; 0.5 means that it is found under half of forest loss

```

```python
def get_main_driver(item):
    classes, counts = item
    main_driver = classes[counts.argmax()]    
    return main_driver
```

```python
def convert_string_to_lists(input_string):
    try:
        lists = ast.literal_eval(input_string)
        if isinstance(lists, tuple) and len(lists) == 2 and all(isinstance(lst, list) for lst in lists):
            return lists
        else:
            return None
    except (SyntaxError, ValueError):
        return None
```

```python
import re
import ast
def modify_string_multiple_items(input_string):
    # Adjusting regex to match and replace patterns for arrays with multiple items
    modified_string = re.sub(r'array\(\[([0-9, ]+)\](, dtype=\w+\d*)?\)', r'[\1]', input_string)
    # Removing any trailing unmatched parentheses
    modified_string = re.sub(r'\)+$', '', modified_string)
    modified_string = modified_string.replace('(','')
    return ast.literal_eval(modified_string)
    #return [eval(item.strip()) for item in  modified_string.split(',')]

# Testing with the new input
output_str_multiple = modify_string_multiple_items('(array([2017, 2020, 2021], dtype=uint16), array([2, 3, 5]))')
output_str_multiple
```

```python
gdf['forest_loss_years'] = gdf['forest_loss_years'].apply(convert_string_to_lists)
gdf['drivers'] = gdf['drivers'].apply(convert_string_to_lists)
```

```python
def filter_forest_loss(item, min_px_loss=50):
    #print(item)
    classes, counts = item
    if isinstance(classes, list):
        classes, counts = np.array(classes), np.array(counts) 
    return counts.sum() >= min_px_loss


def filter_by_driver_ratio(item, min_ratio=0.5):
    try:
        classes, counts = item
        if isinstance(classes, list):
            classes, counts = np.array(classes), np.array(counts)
        main_driver = classes[counts.argmax()]
        ratio = counts[counts.argmax()]/counts.sum()
    except:
        print(item)
    return ratio >= min_ratio
```

```python
gdf_forest_loss_filtered =  gdf.loc[gdf.forest_loss_years.apply(filter_forest_loss)]
gdf_driver_ratio_filtered =  gdf_forest_loss_filtered.loc[gdf_forest_loss_filtered.drivers.apply(filter_by_driver_ratio)]
```

```python
def map_class_indeces_to_drivers(index):
    if index in mapbiomas_classes.keys():
        return mapbiomas_classes[index]
    else:
        return 'other'
```

### Extract a small subset of the data to get started

- for each class, take max_count_per_class (1000) examples, unless there are less available, in which case take all

```python
driver_classes = np.unique(gdf_driver_ratio_filtered.main_driver_name, return_counts=True)[0]
max_count_per_class = 1000

```

```python
driver_classes = [cls for cls in driver_classes if not cls == 'other']
```

```python
driver_classes
```

```python
labels = gdf_driver_ratio_filtered
```

```python
    
def flatten(nested_list):
    return [x for xs in nested_list for x in xs]
```

```python
indices = []
for _class in driver_classes:
    try:
        class_samples = labels.loc[labels.main_driver_name == _class].sample(max_count_per_class, random_state=1).index
    except ValueError:
        class_samples = labels.loc[labels.main_driver_name == _class].index
    indices.append(class_samples.tolist())

indices = flatten(indices)

#return df.loc[indices]
```

```python
subset = labels.loc[indices]
```
