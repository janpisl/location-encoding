```python
import numpy as np
import geopandas as gpd

import matplotlib.pyplot as plt
```

```python
gdf = gpd.read_file('../data/input_datasets/Africa_Masolele/Africa_landuse_following_deforestation_training_Polygon.shp')
gdf = gdf.loc[~gdf.LU.isnull()]
gdf = gdf.loc[(gdf.LU != 7)] #Didn't get the class definition from Robert and there are only 2
```

```python
LU_codes = {
    1 : "large scale cropland",
    2 : "pasture",
    3 : "mining",
    4 : "small scale cropland",
    5 : "tree crops",
    6 : "other tree cover",
    9 : "urban",
    11 : "water",
    12 : "palm oil",
    13 : "rubber",
    14 : "cashew",
    16 : "roads",
    
}

gdf['driver']  = gdf['LU'].map(LU_codes)
```

```python
categories, values = np.unique(gdf['driver'], return_counts=True)
# Plot
plt.figure(figsize=(10, 6))
plt.bar(categories, values, color='skyblue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Labels from Robert')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
