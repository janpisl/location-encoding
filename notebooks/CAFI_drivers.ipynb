```python
import numpy as np
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
```

```python
path = '/home/jan/Documents/EPFL/project_2/data/input_datasets/CAFI_drivers.csv'
df = pd.read_csv(path)
```

```python
df
```

```python
# Convert LON and LAT to a geometry column
geometry = [Point(xy) for xy in zip(df.LON, df.LAT)]

# Create a GeoDataFrame
gdf = gpd.GeoDataFrame(df, geometry=geometry)
```

```python
gdf.columns
```

```python
gdf_valid = gdf.loc[gdf.DriverGroup != '0'].reset_index(drop=True)
```

```python
np.unique(gdf_valid.DriverGroup, return_counts=True)
```

```python
'''
filter:
"countDrivers" = 1  AND "Ref_ArtAg"  = 1

results in 561 items, but they seem like quite clean examples of shifting agriculture
plus approx. 200 more of other drivers when countDrivers == 1


TODO:
    - compare them to the Laos shifting agriculture  THEY MUST LOOK SIMILAR
    - compare to rubber + cocoa examples THEY MUST LOOK DIFFERENT

'''
```

```python


label_mapping = {
    'Rural Complex' : 'shifting agriculture',
    'Industrial Forestry' : 'forest plantation',
    'Artisanal Mine' : 'mining',
    'Industrial Mine' : 'mining',
    'Development' : 'urban',
    'Other' : 'other',
    
    
}
```

```python
gdf_filtered = gdf_valid.loc[gdf_valid.countDrivers == 1].reset_index(drop=True)
gdf_filtered = gdf_filtered.loc[gdf_filtered.DriverGroup.isin(label_mapping.keys())]
```

```python
dataset_prefix = 6000000
```

```python
gdf_filtered['driver'] = gdf_filtered['DriverGroup'].apply(lambda driver: label_mapping[driver])
gdf_filtered['sampleid'] = range(dataset_prefix, dataset_prefix + len(gdf_filtered))
```

```python
gdf_final = gdf_filtered[[ 'driver', 'sampleid',
       'geometry']]
```

```python
gdf_final.to_file('../data/tmp/CAFI.gpkg')
```

```python
pd.DataFrame(gdf_final).to_csv('../data/labels/CAFI.csv', index=False)
```
