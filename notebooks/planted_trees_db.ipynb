```python
import fiona 
import geopandas as gpd
import os 
import pandas as pd
import numpy as np
```

```python
gdb_file = '/home/jan/Documents/EPFL/project_2/data/input_datasets/sdpt_v2_v11282023.gdb/sdpt_v2_v11282023.gdb'
# Get all the layers from the .gdb file 
layers = fiona.listlayers(gdb_file)

i = 0
gdfs = {}
```

```python
skip = True

while i < len(layers):
    
    layer = layers[i]
    i += 1
    if layer == 'chl_plant_v2':
        skip = False
    
    if skip:
        continue
    
    print(layer)
    
    gdf = gpd.read_file(gdb_file,layer=layer)
    print(gdf['originalName'].unique())
    gdfs[layer] = gdf
```

```python
indo = gdfs['idn_plant_v2']
```

```python
#indo.loc['caca' in indo['originalName'].str.lower()]

filtered_indo = indo[indo['originalName'].str.lower().str.contains('caca', na=False)]
```

```python
cocoa_gdfs = []
coffee_gdfs = []


i = 0
```

```python


while i < len(layers):
    
    layer = layers[i]
    i += 1
    
    gdf = gpd.read_file(gdb_file,layer=layer)
    gdfs[layer] = gdf
    print(layer)
    cocoa = gdf[gdf['originalName'].str.lower().str.contains('caca', na=False)]
    coffee = gdf[gdf['originalName'].str.lower().str.contains('coffe', na=False)]

    if not cocoa.empty:
        cocoa_gdfs.append(cocoa)
    if not coffee.empty:
        coffee_gdfs.append(coffee)
    



```

```python
cocoa = pd.concat(cocoa_gdfs)
coffee = pd.concat(coffee_gdfs)
```
