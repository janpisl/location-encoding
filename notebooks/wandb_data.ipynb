```python
import matplotlib.pyplot as plt
from matplotlib.pyplot import figure


import pandas as pd
import numpy as np
import wandb
api = wandb.Api(timeout=20)
```

```python
#wandb_id = 'aagly34c' #cartesian3d
wandb_id = 'y0yvlyz1' #4 images


runs = api.runs(path="janpisl/project_2", filters={'name': {'$in': [wandb_id]}})

print(f"matching runs: {len(runs)}")
```

```python
#For each run, request all data and write to a dict
runs_data = []

for run in iter(runs):
    run_dict = {}
    run_dict['logs'] = run.history()
    run_dict['summary'] = run.summary
    run_dict['config'] = run.config

    runs_data.append(run_dict)
```
